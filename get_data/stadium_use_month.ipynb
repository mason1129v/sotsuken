```python
import pandas as pd
count_all=[]
```

```python
df24=pd.read_csv('cpbl2024.csv')
df24.columns= [
    'No',
    'date',
    'day',
    'stadium',
    'visitor',
    'home',
    'audience'
]
df24.head()
```

```python
stadium=df24['stadium'].unique()
stadium
```

```python
df24['month']=pd.to_datetime(df24['date']).dt.month
df24['month']
df24['year']=pd.to_datetime(df24['date']).dt.year
df24['year']
```

```python
count24=df24.groupby(['year','month','stadium']).size().reset_index(name='count')
count24
```

```python
count_all.append(count24)
count_all
```
