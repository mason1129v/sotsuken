## 試合データと気象データを同じファイルに合わせる

```python
import pandas as pd
```

### 2024年

```python
xz24p=pd.read_csv('気象データ/xinzhuang_2024_prec.csv')
xz24t=pd.read_csv('気象データ/xinzhuang_2024_temp.csv')
tm24p=pd.read_csv('気象データ/tianmu_2024_prec.csv')
tm24t=pd.read_csv('気象データ/tianmu_2024_temp.csv')
ty24p=pd.read_csv('気象データ/taoyuan_2024_prec.csv')
ty24t=pd.read_csv('気象データ/taoyuan_2024_temp.csv')
tk24p=pd.read_csv('気象データ/takao_2024_prec.csv')
tk24t=pd.read_csv('気象データ/takao_2024_temp.csv')
tt24p=pd.read_csv('気象データ/taitung_2024_prec.csv')
tt24t=pd.read_csv('気象データ/taitung_2024_temp.csv')
tp24p=pd.read_csv('気象データ/taipei_2024_prec.csv')
tp24t=pd.read_csv('気象データ/taipei_2024_temp.csv')
tn24p=pd.read_csv('気象データ/tainan_2024_prec.csv')
tn24t=pd.read_csv('気象データ/tainan_2024_temp.csv')
tc24p=pd.read_csv('気象データ/taichung_2024_prec.csv')
tc24t=pd.read_csv('気象データ/taichung_2024_temp.csv')
hl24p=pd.read_csv('気象データ/hualien_2024_prec.csv')
hl24t=pd.read_csv('気象データ/hualien_2024_temp.csv')
dl24p=pd.read_csv('気象データ/douliu_2024_prec.csv')
dl24t=pd.read_csv('気象データ/douliu_2024_temp.csv')
cy24p=pd.read_csv('気象データ/chiayi_2024_prec.csv')
cy24t=pd.read_csv('気象データ/chiayi_2024_temp.csv')
```

```python
df24=pd.read_csv('cpbl2024.csv')
df24.columns= [
    'No',
    'date',
    'day',
    'stadium',
    'visitor',
    'home',
    'audience'
]
df24.head()
```

```python
df24['year']=pd.to_datetime(df24['date']).dt.year
print(df24['year'])
df24['month']=pd.to_datetime(df24['date']).dt.month
print(df24['month'])
df24['d']=pd.to_datetime(df24['date']).dt.day
print(df24['d'])
```

```python
df24.head()
```

```python
df_list = []

stadium_weather_map = {
    '新莊': (xz24p, xz24t),
    '天母': (tm24p, tm24t),
    '樂天桃園': (ty24p, ty24t),
    '澄清湖': (tk24p, tk24t),
    '台東': (tt24p, tt24t),
    '台北大巨蛋': (tp24p, tp24t),
    '台南': (tn24p, tn24t),
    '洲際': (tc24p, tc24t),
    '花蓮': (hl24p, hl24t),
    '斗六': (dl24p, dl24t),
    '嘉義市': (cy24p, cy24t)
}

for stadium_name, (df_p, df_t) in stadium_weather_map.items():
    df_sub = df24[df24['stadium'].astype(str).str.strip() == stadium_name].copy()
    
    if df_sub.empty:
        continue

    df_sub['month'] = df_sub['month'].astype(int)
    df_sub['d'] = df_sub['d'].astype(int)

    pv = df_p.melt(id_vars=['日/月'], var_name='month', value_name='prec')
    pv = pv[pv['日/月'].astype(str).str.isnumeric()].copy()
    pv['month'] = pv['month'].astype(int)
    pv['d'] = pv['日/月'].astype(int)

    tv = df_t.melt(id_vars=['日/月'], var_name='month', value_name='temp')
    tv = tv[tv['日/月'].astype(str).str.isnumeric()].copy()
    tv['month'] = tv['month'].astype(int)
    tv['d'] = tv['日/月'].astype(int)

    df_sub = pd.merge(df_sub, pv[['month', 'd', 'prec']], on=['month', 'd'], how='left')
    df_sub = pd.merge(df_sub, tv[['month', 'd', 'temp']], on=['month', 'd'], how='left')

    df_list.append(df_sub)


df24_all_weather = pd.concat(df_list, ignore_index=True)

df24_all_weather['date_dt'] = pd.to_datetime(df24_all_weather['date'])
df24_all_weather = df24_all_weather.sort_values('date_dt').drop(columns=['date_dt']).reset_index(drop=True)

df24_all_weather.head(10)
```

```python
select_cols = ['No', 'date', 'day', 'stadium', 'visitor', 'home', 'audience', 'prec', 'temp']
df24_all = df24_all_weather[select_cols].copy()

df24_all.head()
```

```python
rename_map = {
    'No': '試合番号',
    'date': '年月日',
    'day': '曜日',
    'stadium': '球場',
    'visitor': 'ビジター',
    'home': 'ホーム',
    'audience': '観客数',
    'prec': '降水量',
    'temp': '気温'
}

df = df24_all.rename(columns=rename_map)
df.head()
```

```python
df.to_csv('games_weather_2024.csv', index=False, encoding='utf-8-sig')
```
