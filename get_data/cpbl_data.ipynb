##　試合データ収集

#### 2024のデータ

```python
import requests
from bs4 import BeautifulSoup
import pandas as pd

#2024年のデータ
url = "https://zxc22.idv.tw/cpbl35/allgame.asp?PgSz=360"

headers = {
    "User-Agent": "Mozilla/5.0"
}

r = requests.get(url, headers=headers)

print(r.status_code) 
# print(r.text[1810:])
# print(r.content)
```

```python
soup = BeautifulSoup(r.text[1810:], 'html.parser')

tds = soup.find_all('td')
games=[]
game=[]
c=0
for td in tds:
    game.append(td.text.strip()+" ")
    c+=1
    if(c==13):
        c=0
        games.append(game)
        game=[]
# print(games)
```

```python
games_data=[]
game_data=[]
# No date day stadium away home audience
for g in games:
    game_data.append(g[0])
    game_data.append(g[1][:-4])
    game_data.append(g[1][-3:-2])
    game_data.append(g[2])
    game_data.append(g[3])
    game_data.append(g[4])
    game_data.append(g[7])
    games_data.append(game_data)
    game_data=[]

# print(len(games_data))
# print(games_data)
```

```python
columns = [
    '試合番号',
    '年月日',
    '曜日',
    '球場',
    'ビジター',
    'ホーム',
    '観客数'
]

df = pd.DataFrame(games_data, columns=columns)

df.to_csv('cpbl2024.csv', index=False, encoding='utf-8-sig')
```

```python
gd=[]
for i in range(360):
    gd.append(games_data[i][3])
print(set(gd))
```
